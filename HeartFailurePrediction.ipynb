# IMPORT LIBRARIES


```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt 
```

# Importing Dataset


```python
dataset = pd.read_csv('heart.csv')
x = dataset.iloc[:, :-1].values
y = dataset.iloc[:, -1].values
print(x)
print(y)
```

# DATA PREPROCESSING

```python
from sklearn.preprocessing import LabelEncoder

le1 = LabelEncoder()
le2 = LabelEncoder()
le6 = LabelEncoder()
le8 = LabelEncoder()
le10 = LabelEncoder()
x[:, 1] = le1.fit_transform(x[:, 1])
x[:, 2] = le2.fit_transform(x[:, 2])
x[:, 6] = le6.fit_transform(x[:, 6])
x[:, 8] = le8.fit_transform(x[:, 8])
x[:, 10] = le10.fit_transform(x[:, 10])
```

```python
print(x)
```

# Splitting Datatset into Training and Test set


```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2, random_state = 0)
```

```python
print(x_train)
print(x_test)
print(y_train)
print(y_test)
```

# Feature Scaling

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
x_train = sc.fit_transform(x_train)
x_test = sc.fit_transform(x_test)
```

```python
print(x_test)
```

# Training Dataset

```python
#Logistic Regression
from sklearn.linear_model import LogisticRegression
model_logistic = LogisticRegression(random_state = 0)
model_logistic.fit(x_train, y_train)
y_pred_logistic = model_logistic.predict(x_test)
```

# Testing Accuracy

```python
from sklearn.metrics import accuracy_score
accuracy_logistic = accuracy_score(y_test, y_pred_logistic)
print("Accuracy of Logistic Regression: {:.2f}%".format(accuracy_logistic * 100))
```

# SVM (Support vector Machine)

```python
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score
model_svm = SVC()
model_svm.fit(x_train, y_train)
y_pred_svm = model_svm.predict(x_test)
accuracy_svm = accuracy_score(y_test, y_pred_svm)
print("Accuracy of SVM: {:.2f}%".format(accuracy_svm * 100))
```

# k Neighbors

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
model_knn = KNeighborsClassifier(n_neighbors = 8)
model_knn.fit(x_train, y_train)
y_pred_knn = model_knn.predict(x_test)
accuracy_knn = accuracy_score(y_test, y_pred_knn)
print("Accuracy of KNN: {:.2f}%".format(accuracy_knn * 100))
```

# Decision tree

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
model_dt = DecisionTreeClassifier()
model_dt.fit(x_train, y_train)  
y_pred_dt = model_dt.predict(x_test)
accuracy_dt = accuracy_score(y_test, y_pred_dt)
print("Accuracy of Decision Tree: {:.2f}%".format(accuracy_dt * 100))
```

# Random forest

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
model_rf = RandomForestClassifier(n_estimators = 200)
model_rf.fit(x_train, y_train)
y_pred_rf = model_rf.predict(x_test)
accuracy_rf = accuracy_score(y_test, y_pred_rf)
print("Accuracy of Random Forest: {:.2f}%".format(accuracy_rf * 100))
```

# XGBoost

```python
from xgboost import XGBClassifier
model_xgb = XGBClassifier()
model_xgb.fit(x_train, y_train)
y_pred_xgb = model_xgb.predict(x_test)
accuracy_xgb = accuracy_score(y_test, y_pred_xgb)
print("Accuracy of XGBoost: {:.2f}%".format(accuracy_xgb * 100))
```
